```python
import numpy as np
import pandas as pd

print("Numpy version:",np.__version__)
print("Pandas version:",pd.__version__)
```

```python
marks = np.array([78,85,92,67,88])

print("Original:", marks)
print("Add 5:",marks+5)
print("Multiply by 2:",marks*2)
print("Divide by 2:",marks/2)
print("Square by 2:",marks**2)
```

```python
print("Sum:",marks.sum())
print("Mean:",marks.mean())
print("Max:",marks.max())
print("Min:",marks.min())
print("Standard Deviation:",marks.std())
```

```python
print("First mark",marks[0])
print("Second mark",marks[1])
```

```python
std_marks = np.array([
    [78, 85, 92],
    [67, 88, 75],
    [90, 95, 89]
])

print(std_marks)
print("Shape:",std_marks.shape)
```

```python
print("Mean of all values:", std_marks.mean())
print("Mean of each column:", std_marks.mean(axis=0))
print("Mean of each row:", std_marks.mean(axis=1))
```
